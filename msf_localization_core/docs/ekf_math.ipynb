```python
import sympy as sp
```

```python
# =========================
# =========================
# Symbols
# =========================
# =========================

# Time
dt = sp.symbols('dt')

# Position
p = sp.Matrix([
    sp.Symbol('x'),
    sp.Symbol('y'),
    sp.Symbol('z')
])

# Orientation
# roll(ϕ), pitch(θ), yaw(ψ)
phi, theta, psi = sp.symbols('phi theta psi')

orientation = sp.Matrix([
    phi,
    theta,
    psi
])

# Linear Velocity
v = sp.Matrix([
    sp.Symbol('vx'),
    sp.Symbol('vy'),
    sp.Symbol('vz')
])

# Angular Velocity measured by IMU
w = sp.Matrix([
    sp.Symbol('wx'),
    sp.Symbol('wy'),
    sp.Symbol('wz')
])

# Linear Acceleration measured by IMU
a = sp.Matrix([
    sp.Symbol('ax'),
    sp.Symbol('ay'),
    sp.Symbol('az')
])

# Gravity
g = sp.symbols('g')
gravity = sp.Matrix([
    0,
    0,
    -g
])

# Rotation matrix R
Rx = sp.Matrix([
    [1, 0, 0],
    [0, sp.cos(phi), -sp.sin(phi)],
    [0, sp.sin(phi),  sp.cos(phi)]
])

Ry = sp.Matrix([
    [ sp.cos(theta), 0, sp.sin(theta)],
    [0,              1, 0],
    [-sp.sin(theta), 0, sp.cos(theta)]
])

Rz = sp.Matrix([
    [sp.cos(psi), -sp.sin(psi), 0],
    [sp.sin(psi),  sp.cos(psi), 0],
    [0,             0,            1]
])

# ZYX convention
R = Rz * Ry * Rx

# acceleration_world = R * a + gravity

# roll(ϕ), pitch(θ), yaw(ψ)
# phi,     theta,    psi
E = sp.Matrix([
    [1, sp.sin(phi) * sp.tan(theta), sp.cos(phi) * sp.tan(theta)],
    [0, sp.cos(phi), -sp.sin(phi)],
    [0, sp.sin(phi) / sp.cos(theta), sp.cos(phi) / sp.cos(theta)]
])

# =========================
# =========================
# Motion Model
# =========================
# =========================

# -------------------------
# position equation
# -------------------------
position_next = p + R * v * dt + sp.Rational(1, 2) * R * a * dt**2

# -------------------------
# orientation equation
# -------------------------
orientation_next = orientation + E * w * dt

# -------------------------
# velocity equation
# -------------------------
velocity_next = v + a * dt

# =========================
# =========================
# state transition function
# =========================
# =========================

# state
state = sp.Matrix([
    p[0], p[1], p[2],
    phi, theta, psi,
    v[0],v[1], v[2]
])

# predicted state
f = sp.Matrix.vstack(
    position_next,
    orientation_next,
    velocity_next
)

# Jacobian
F = f.jacobian(state)
```
